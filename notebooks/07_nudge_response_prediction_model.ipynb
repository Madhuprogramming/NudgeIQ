```python
# ============================================================
# NUDGEIQ - NOTEBOOK 07
# INITIALIZE DATA
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("Starting NudgeIQ...")

# Find project root
project_root = Path.cwd().parent

print("Current directory:", Path.cwd())
print("Project root:", project_root)

# Load the behavioral intelligence dataset
data_path = (
    project_root
    / "data"
    / "processed"
    / "behavioral_intelligence.csv"
)

print("Looking for:")
print(data_path)

# Check whether file exists
if not data_path.exists():
    raise FileNotFoundError(
        f"\nFile not found:\n{data_path}\n\n"
        "Please check that your notebook is inside the NudgeIQ/notebooks folder."
    )

# Load dataset
data = pd.read_csv(data_path)

print("\nSUCCESS!")
print("Dataset shape:", data.shape)

print("\nFirst 5 rows:")
display(data.head())

print("\nColumns:")
print(data.columns.tolist())
```

```python
# ============================================================
# CHECK TARGET
# ============================================================

print("Checking deposit column...")

if "deposit" not in data.columns:
    raise KeyError(
        "The column 'deposit' does not exist in this dataset."
    )

print("\nDeposit values:")
print(data["deposit"].value_counts(dropna=False))
```

```python
# ============================================================
# CELL 2 — TARGET CHECK
# ============================================================

print("=" * 60)
print("TARGET VARIABLE CHECK")
print("=" * 60)

print("\nDeposit values:")
print(data["deposit"].value_counts(dropna=False))

print("\nDeposit percentages:")
print(
    data["deposit"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# CELL 3 — CREATE BINARY TARGET
# ============================================================

data["target"] = (
    data["deposit"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("yes")
    .astype(int)
)

print("Target created successfully.")

print("\nTarget counts:")
print(data["target"].value_counts())

print("\nTarget percentages:")
print(
    data["target"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# CELL 4 — DEFINE BASELINE PREDICTORS
# ============================================================

feature_columns = [
    # Customer profile
    "age",
    "job",
    "marital",
    "education",

    # Financial position
    "default",
    "balance",
    "housing",
    "loan",

    # Contact / campaign information
    "contact",
    "campaign",
    "pdays",
    "previous",
    "poutcome"
]

X = data[feature_columns].copy()
y = data["target"].copy()

print("=" * 60)
print("BASELINE MODEL DATA")
print("=" * 60)

print("\nX shape:", X.shape)
print("y shape:", y.shape)

print("\nPredictor columns:")
for column in feature_columns:
    print(" -", column)

print("\nMissing values:")
print(X.isnull().sum().sum())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# ============================================================
# CELL 5 — TRAIN / TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("=" * 60)
print("TRAIN / TEST SPLIT")
print("=" * 60)

print("\nTraining rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining target:")
print(
    y_train.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nTesting target:")
print(
    y_test.value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# CELL 6 — PREPROCESSING
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Categorical variables
categorical_features = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "poutcome"
]

# Numerical variables
numerical_features = [
    "age",
    "balance",
    "campaign",
    "pdays",
    "previous"
]

print("=" * 60)
print("PREPROCESSING SETUP")
print("=" * 60)

print("\nCategorical features:")
for col in categorical_features:
    print(" -", col)

print("\nNumerical features:")
for col in numerical_features:
    print(" -", col)

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)

print("\nPreprocessor created successfully.")
```

```python
# ============================================================
# CELL 7 — LOGISTIC REGRESSION
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=2000,
                random_state=42
            )
        )
    ]
)

# Train
logistic_model.fit(
    X_train,
    y_train
)

print("=" * 60)
print("LOGISTIC REGRESSION")
print("=" * 60)

print("Model trained successfully.")
```

```python
# ============================================================
# CELL 8 — LOGISTIC REGRESSION EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Predictions
logistic_prediction = logistic_model.predict(X_test)

# Probability of positive class
logistic_probability = (
    logistic_model.predict_proba(X_test)[:, 1]
)

# Metrics
logistic_accuracy = accuracy_score(
    y_test,
    logistic_prediction
)

logistic_precision = precision_score(
    y_test,
    logistic_prediction
)

logistic_recall = recall_score(
    y_test,
    logistic_prediction
)

logistic_f1 = f1_score(
    y_test,
    logistic_prediction
)

logistic_auc = roc_auc_score(
    y_test,
    logistic_probability
)

print("=" * 60)
print("LOGISTIC REGRESSION — MODEL EVALUATION")
print("=" * 60)

print(f"\nAccuracy : {logistic_accuracy:.4f}")
print(f"Precision: {logistic_precision:.4f}")
print(f"Recall   : {logistic_recall:.4f}")
print(f"F1 Score : {logistic_f1:.4f}")
print(f"ROC-AUC  : {logistic_auc:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        logistic_prediction
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        logistic_prediction
    )
)
```

```python
# ============================================================
# CELL 14 — GOLD NUDGE PRIORITY
# ============================================================

def assign_gold_priority(row):

    priority = str(row["Priority"]).strip().lower()
    persona = str(row["Persona"]).strip()

    # High-priority customers
    if priority == "high":
        return "HIGH"

    # Medium-priority customers
    elif priority == "medium":
        return "MEDIUM"

    # Low-priority customers
    elif priority == "low":
        return "LOW"

    # Safety fallback
    else:
        return "LOW"


nudgeiq_gold["Gold_Nudge_Priority"] = (
    nudgeiq_gold.apply(
        assign_gold_priority,
        axis=1
    )
)

print("=" * 70)
print("GOLD NUDGE PRIORITY")
print("=" * 70)

print("\nGold Nudge Priority distribution:")
print(
    nudgeiq_gold["Gold_Nudge_Priority"]
    .value_counts()
)

print("\nGold Nudge Priority percentage:")
print(
    nudgeiq_gold["Gold_Nudge_Priority"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# CELL 29 — GOLD OPPORTUNITY QUADRANT
# ============================================================

# ------------------------------------------------------------
# Determine data-driven thresholds
# ------------------------------------------------------------

balance_threshold = (
    final_gold_nudgeiq["balance"]
    .median()
)

priority_threshold = (
    final_gold_nudgeiq["Nudge_Priority_Score"]
    .median()
)

print("=" * 70)
print("GOLD OPPORTUNITY QUADRANT")
print("=" * 70)

print(
    f"\nFinancial capacity threshold (median balance): "
    f"₹{balance_threshold:,.2f}"
)

print(
    f"Behavioral priority threshold (median Nudge Score): "
    f"{priority_threshold:.3f}"
)


# ------------------------------------------------------------
# Assign quadrant
# ------------------------------------------------------------

def assign_gold_quadrant(row):

    high_balance = (
        row["balance"] >= balance_threshold
    )

    high_priority = (
        row["Nudge_Priority_Score"] >= priority_threshold
    )

    if high_balance and high_priority:
        return "STRATEGIC GOLD TARGET"

    elif not high_balance and high_priority:
        return "PRIORITY GOLD TARGET"

    elif high_balance and not high_priority:
        return "NURTURE GOLD OPPORTUNITY"

    else:
        return "GOLD AWARENESS"

    
final_gold_nudgeiq["Gold_Opportunity_Quadrant"] = (
    final_gold_nudgeiq.apply(
        assign_gold_quadrant,
        axis=1
    )
)


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

quadrant_summary = (
    final_gold_nudgeiq
    .groupby("Gold_Opportunity_Quadrant")
    .agg(
        Customers=("Customer_ID", "count"),
        Total_Balance=("balance", "sum"),
        Average_Balance=("balance", "mean"),
        Average_Nudge_Score=(
            "Nudge_Priority_Score",
            "mean"
        )
    )
    .reset_index()
)

quadrant_summary["Customer_%"] = (
    quadrant_summary["Customers"]
    / len(final_gold_nudgeiq)
    * 100
).round(2)

quadrant_summary["Balance_%"] = (
    quadrant_summary["Total_Balance"]
    / final_gold_nudgeiq["balance"].sum()
    * 100
).round(2)

quadrant_summary["Total_Balance"] = (
    quadrant_summary["Total_Balance"]
    .round(2)
)

quadrant_summary["Average_Balance"] = (
    quadrant_summary["Average_Balance"]
    .round(2)
)

quadrant_summary["Average_Nudge_Score"] = (
    quadrant_summary["Average_Nudge_Score"]
    .round(3)
)

display(
    quadrant_summary.sort_values(
        "Total_Balance",
        ascending=False
    )
)
```
