```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

behavior = pd.read_csv(
    r"D:\NudgeIQ\data\processed\behavioral_features.csv"
)

behavior.head()
```

```python
behavioral_constructs = [
    "Financial_Stability",
    "Contact_Intensity",
    "Relationship_History",
    "Investment_Experience",
    "Responsiveness",
    "Investment_Readiness"
]
```

```python
behavior["Financial_Index"] = behavior["Financial_Stability"]
```

```python
print(behavior.columns.tolist())
```

```python
scaler = StandardScaler()

behavior["Engagement_Index"] = scaler.fit_transform(
    behavior[["Engagement_Index"]]
)
```

```python
behavior["Investor_Profile_Index"] = (
      0.40 * behavior["Investment_Readiness"]
    + 0.35 * behavior["Risk_Appetite"]
    + 0.25 * behavior["Investment_Experience"]
)
```

```python
behavior["Investor_Profile_Index"] = scaler.fit_transform(
    behavior[["Investor_Profile_Index"]]
)
```

```python
behavior["Behavioral_Intelligence_Score"] = (
      0.40 * behavior["Financial_Index"]
    + 0.30 * behavior["Engagement_Index"]
    + 0.30 * behavior["Investor_Profile_Index"]
)
```

```python
behavior["Behavioral_Intelligence_Score"] = scaler.fit_transform(
    behavior[["Behavioral_Intelligence_Score"]]
)
```

```python
behavior[[
    "Financial_Index",
    "Engagement_Index",
    "Investor_Profile_Index",
    "Behavioral_Intelligence_Score"
]].describe()
```

```python
constructs = [
    "Financial_Stability",
    "Investment_Readiness",
    "Digital_Engagement",
    "Risk_Appetite",
    "Loyalty_Potential",
    "Responsiveness",
    "Investment_Experience"
]

construct_corr = behavior[constructs].corr().round(2)

print(construct_corr)
```

```python
financial_corr = (
    construct_corr["Financial_Stability"]
    .drop("Financial_Stability")
    .sort_values(ascending=False)
)

print("\nFinancial_Stability relationships")
print("-" * 40)
print(financial_corr)
```

```python
behavior[[
    "Financial_Index",
    "Engagement_Index",
    "Investor_Profile_Index",
    "Behavioral_Intelligence_Score"
]].corr().round(2)
```

```python
behavior.to_csv(
    r"D:\NudgeIQ\data\processed\behavioral_intelligence.csv",
    index=False
)

print("✅ behavioral_intelligence.csv created successfully!")
```
