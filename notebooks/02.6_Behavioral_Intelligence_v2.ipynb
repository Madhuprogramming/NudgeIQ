```python
behavior = pd.read_csv(
    r"D:\NudgeIQ\data\processed\behavioral_data.csv"
)

print("Before:", behavior.shape)
print("Before columns:", len(behavior.columns))

behavior = create_behavioral_features(behavior)

print("After:", behavior.shape)
print("After columns:", len(behavior.columns))

print(behavior.columns.tolist())
```

```python
behavior["Financial_Index"] = behavior["Financial_Stability"]
```

```python
behavior["Engagement_Index"] = (
    0.60 * behavior["Contact_Intensity"]
    + 0.40 * behavior["Relationship_History"]
)
```

```python
behavior["Investor_Profile_Index"] = (
    behavior["Investment_Readiness"]
)
```

```python
from sklearn.preprocessing import StandardScaler

index_cols = [
    "Financial_Index",
    "Engagement_Index",
    "Investor_Profile_Index"
]

behavior[index_cols] = StandardScaler().fit_transform(
    behavior[index_cols]
)
```

```python
behavior["Behavioral_Intelligence_Score"] = (
      0.35 * behavior["Financial_Index"]
    + 0.35 * behavior["Engagement_Index"]
    + 0.30 * behavior["Investor_Profile_Index"]
)

behavior["Behavioral_Intelligence_Score"] = StandardScaler().fit_transform(
    behavior[["Behavioral_Intelligence_Score"]]
)
```

```python
indices = [
    "Financial_Index",
    "Engagement_Index",
    "Investor_Profile_Index",
    "Behavioral_Intelligence_Score",
]

behavior[indices].describe()
```

```python
behavior.to_csv(
    r"D:\NudgeIQ\data\processed\behavioral_intelligence.csv",
    index=False
)

print("✅ behavioral_intelligence.csv created successfully")
```
