```python
# ============================================================
# NOTEBOOK 06 — NUDGE IMPACT / UPLIFT MODEL
# ============================================================

import pandas as pd
from pathlib import Path

project_root = Path.cwd().parent

behavior_path = (
    project_root
    / "data"
    / "processed"
    / "behavioral_data.csv"
)

behavior_raw = pd.read_csv(behavior_path)

print("Dataset shape:", behavior_raw.shape)

print("\nColumns:")
print(behavior_raw.columns.tolist())

print("\nOutcome distribution:")
print(behavior_raw["deposit"].value_counts())

print("\nOutcome percentage:")
print(
    behavior_raw["deposit"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# CHECK ROW ALIGNMENT BEFORE MERGING
# ============================================================

final_output_path = (
    project_root
    / "data"
    / "processed"
    / "final_nudgeiq_output.csv"
)

nudge_output = pd.read_csv(final_output_path)

print("Original behavioral data:", behavior_raw.shape)
print("Final NudgeIQ output:", nudge_output.shape)

# Compare identifying raw columns
check_columns = [
    "age",
    "job",
    "marital",
    "education",
    "balance"
]

print("\nRow alignment check:")

for col in check_columns:
    matches = (
        behavior_raw[col].astype(str).values
        == nudge_output[col].astype(str).values
    ).all()

    print(f"{col}: {'✓ MATCH' if matches else '✗ DIFFERENT'}")
```
